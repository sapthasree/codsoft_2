```python
# import necessary libraries
import pandas as pd
```

```python
# load data
movies = pd.read_csv('movie.csv', encoding='latin1')
movies.head()
```

```python
# NA values
movies.isna()
```

```python
# handling missing values
data = movies.dropna(subset=['Year','Duration','Genre','Rating','Votes','Director','Actor 1','Actor 2','Actor 3'])
data
```

```python
# checking if new data have missing values
data.isna()
```

```python
# label encoder
from sklearn.preprocessing import LabelEncoder
for col in ['Genre','Director','Actor 1','Actor 2','Actor 3']:
    data.loc[:,col] = LabelEncoder().fit_transform(data[col])

data.loc[:,'Duration'] = data['Duration'].astype(str).str.replace(' min','').astype(float)

data = data.copy()
data.loc[:,'Year'] = pd.to_numeric(data['Year'].astype(str).str.extract(r'(\d{4})')[0], errors='coerce')
```

```python
data
```

```python
# initialising predictor and response variables
x = data.drop(['Name','Rating'],axis=1)
y = data.Rating
```

```python
# training and testing sets
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# model building
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
# Clean the data by removing commas and converting to float
x_train = x_train.replace(',', '', regex=True).infer_objects(copy=False).astype(float)
y_train = y_train.replace(',', '', regex=True).infer_objects(copy=False).astype(float)
x_test = x_test.replace(',', '', regex=True).infer_objects(copy=False).astype(float)
y_test = y_test.replace(',', '', regex=True).infer_objects(copy=False).astype(float)


# fitting the data to the model
model.fit(x_train,y_train)
```

```python
# prediction
y_pred = model.predict(x_test)
```

```python
# computing error
from sklearn.metrics import mean_absolute_error
mae = mean_absolute_error(y_test,y_pred)
print('Mean Absolute Error:',mae)
```

```python
# resultant matrix
import numpy as np
result_matrix = np.column_stack((y_test, y_pred))
print('  True         Predicted')
print(result_matrix)
```
